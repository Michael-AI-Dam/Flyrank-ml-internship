```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd

# Load starter dataset
df = pd.read_csv("https://raw.githubusercontent.com/Michael-AI-Dam/Flyrank-ml-internship/main/data/raw/content_refresh_anonymized.csv")

# See what we're working with
print("Shape:", df.shape)
print("Columns:", list(df.columns))
```

```python
ctr_col = "ctr"
impressions_col = "impressions_last_30d"

total_rows = len(df)
avg_ctr = df[ctr_col].mean()
high_impr_low_ctr = df[(df[impressions_col] > 1000) & (df[ctr_col] < 0.05)].shape[0]

print("Total rows:", total_rows)
print("Average CTR:", round(avg_ctr * 100, 2), "%")
print("High impressions + low CTR rows:", high_impr_low_ctr)
```
